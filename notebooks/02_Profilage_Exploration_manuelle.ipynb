# Analyse Manuelle

*Changer type des attributs*

```python
# Convert dates
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

*Calculer les valeurs manquantes*

```python
# Missing values
missing_df = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": df.isna().mean() * 100
}).sort_values("missing_percent", ascending=False)

missing_df
```

*Duplicates patients*

```python
duplicates = df[df.duplicated(
    subset=["Name", "Date of Admission"], keep=False
)]
duplicates.shape
```

```python
duplicates_sorted = (
    df[df.duplicated(subset=["Name", "Date of Admission"], keep=False)]
    .sort_values(by=["Name", "Date of Admission"])
)

duplicates_sorted.head(20)
```

*Patients avec trop de réadmissions rapprochées*

si Name apparaît plusieurs fois sur une courte période

```python
df_sorted = df.sort_values(["Name", "Date of Admission"])

df_sorted["Prev_Discharge"] = df_sorted.groupby("Name")["Discharge Date"].shift(1)

df_sorted["Days_Between"] = ( df_sorted["Date of Admission"] - df_sorted["Prev_Discharge"] ).dt.days

readmission_anomalies = df_sorted[df_sorted["Days_Between"] < 7]
readmission_anomalies.shape
```

```python
result = readmission_anomalies.sort_values(
    ["Name", "Date of Admission"]
)
cols_to_view = [
    "Name",
    "Date of Admission",
    "Discharge Date",
    "Prev_Discharge",
    "Days_Between",
    "Medical Condition",
    "Admission Type",
    "Hospital"
]

result[cols_to_view].head(20)

```

*La Cohérence Clinique (Médicaments vs Pathologie)*

On va chercher à Vérifier si les médicaments prescrits sont cohérents avec la maladie.
*Exemple de cas*  : On ne donne pas de la Pénicilline pour de l'Arthrite ou de l'Asthme (c'est un antibiotique). On ne donne pas d'Aspirine pour soigner un Cancer.

```python

# Vérifier quels médicaments sont donnés pour l'Asthme
asthma_meds = df[df["Medical Condition"] == "Asthma"]["Medication"].value_counts()
print("Médicaments prescrits pour l'Asthme :\n", asthma_meds)

# Si on voit "Penicillin" ou "Lipitor" pour de l'Asthme, c'est une incohérence métier majeure.
```

*La Logique Assurantielle (Age vs Assurance)*

Aux USA (contexte probable vu les noms d'assurances), Medicare est strictement réservé aux personnes de 65 ans et plus (sauf rares exceptions de handicap).

L'anomalie probable : Des patients jeunes (ex: 20 ans) couverts par Medicare.

```python
# Cohérence Age vs Medicare
medicare_anomaly = df[
    (df["Insurance Provider"] == "Medicare") &
    (df["Age"] < 65)
]
print(f"Nombre de patients < 65 ans sous Medicare : {medicare_anomaly.shape[0]}")
```

*La Gestion des Chambres (Collisions Spatiales)*

Vérification de l'occupation temporelle des chambres.

Le problème physique : Une même chambre (ex: 101) ne peut pas être occupée par deux patients différents en même temps.

```python
# Collision de chambres : Deux patients admis dans la même chambre le même jour
room_collision = df[df.duplicated(subset=["Room Number", "Date of Admission"], keep=False)]
print(f"Collisions de chambres détectées : {room_collision.shape[0]}")
```

*La Qualité des Textes (Problèmes de format)*

casse aléatoire : Compter combien de noms contiennent des majuscules au milieu du mot.

 les noms d'hôpitaux business : Chercher les suffixes "LLC" ou "Inc" qui ne devraient pas être là pour un hôpital.

```python
# Détection de casse irrégulière (lettre majuscule au milieu d'un nom)
# Regex : une minuscule suivie d'une majuscule
irregular_case = df[df["Name"].str.contains(r'[a-z][A-Z]', regex=True)]
print(f"Noms avec casse mixte suspecte : {irregular_case.shape[0]}")

# Détection de suffixes commerciaux dans les hôpitaux
commercial_hospitals = df[df["Hospital"].str.contains(r' LLC| Inc| Ltd', regex=True)]
print(f"Hôpitaux avec noms commerciaux (LLC, Inc...) : {commercial_hospitals.shape[0]}")
```
