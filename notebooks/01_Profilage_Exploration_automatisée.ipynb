```python
import pandas as pd
import sweetviz as sv
from sqlalchemy import create_engine, inspect


# 1. CONFIGURATION DE LA CONNEXION (Docker)
host = 'mariadb'
user = 'root'
password = 'root'
port = '3306'
db_name = 'data_quality_db'

engine = create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{db_name}")


# 2. INGESTION ET CHARGEMENT (Architecture Bronze)
try:
    print(f"Connexion à MariaDB ({db_name})...")
    
    # Vérifier si la table 'healthcare_bronze' existe déjà
    inspector = inspect(engine)
    if not inspector.has_table('healthcare_bronze'):
        print("La table Bronze n'existe pas. Création en cours à partir du CSV...")
        # On charge le CSV une seule fois pour initialiser la base
        df_csv = pd.read_csv("/home/jovyan/work/data/raw/healthcare_dataset.csv")
        df_csv.to_sql('healthcare_bronze', con=engine, if_exists='replace', index=False)
        print("Table 'healthcare_bronze' créée !")
    
    # Lire les données depuis la Base de Données (Le vrai flux !)
    print("Chargement des données depuis la table SQL 'healthcare_bronze'...")
    df = pd.read_sql("SELECT * FROM healthcare_bronze", con=engine)
    print(f"Données chargées avec succès : {df.shape}")


# 3. PROFILAGE AUTOMATISÉ (Sweetviz)
    print("Lancement de l'analyse Sweetviz...")
    report = sv.analyze(df)
    report.show_html('Rapport_Audit_Quality.html')
    print("Rapport généré : Rapport_Audit_Quality.html")

except Exception as e:
    print(f"Erreur : {e}")
```
