```python
import pandas as pd
import sweetviz as sv
import os

# 1. Vérification du chemin (pour être sûr)
file_path = '../data/raw/healthcare_dataset.csv'

if os.path.exists(file_path):
    print("Fichier trouvé ! Chargement en cours...")
    
    # 2. Chargement des données
    df = pd.read_csv(file_path)
    
    # 3. Génération du rapport (Le Scanner)
    report = sv.analyze(df)
    
    # 4. Sauvegarde du rapport en HTML
    report.show_html('Rapport_Audit_Quality.html')
    print("Rapport généré avec succès !")
    
else:
    print(f"Erreur : Le fichier n'est pas trouvé à cet endroit : {file_path}")
    print("Vérifie que tu as bien mis ton CSV dans le dossier 'data/raw' de ton PC.")
```
